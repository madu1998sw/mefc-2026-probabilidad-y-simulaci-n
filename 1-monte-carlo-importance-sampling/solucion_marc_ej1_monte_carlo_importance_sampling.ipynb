```python
import numpy as np

# Semilla fija para que te coincidan los números exactos
np.random.seed(42)
N = 200

# Función original g(x) del enunciado
def g(x):
    return np.cos(np.pi * (x**2) / 2)

# Densidad propuesta f_tilde(x) calculada en el apartado 1
def f_tilde(x):
    return 1.5 * (1 - x**2)
```

```python
# Muestreo con distribución Uniforme U(0,1)
u_direct = np.random.uniform(0, 1, N)
eval_direct = g(u_direct)

integral_direct = np.mean(eval_direct)
variance_direct = np.var(eval_direct, ddof=1) / N

print("--- MÉTODO DIRECTO ---")
print(f"Estimación de la Integral: {integral_direct:.6f}")
print(f"Varianza del estimador: {variance_direct:.6e}")
```

```python
# Generación usando la transformada inversa y Cardano
u_importance = np.random.uniform(0, 1, N)
x_importance = 2 * np.cos(np.arccos(-u_importance) / 3)

# Cálculo del estimador ponderado por los pesos
pesos = g(x_importance) / f_tilde(x_importance)

integral_importance = np.mean(pesos)
variance_importance = np.var(pesos, ddof=1) / N

print("--- MUESTREO POR IMPORTANCIA ---")
print(f"Estimación de la Integral: {integral_importance:.6f}")
print(f"Varianza del estimador: {variance_importance:.6e}")
print(f"Reducción de varianza: {variance_direct / variance_importance:.2f} veces menor")
```
